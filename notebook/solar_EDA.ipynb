## Summary
The solar eclipse dataset contains 11,898 records with 15 columns, spanning catalog data 
on eclipse type, timing, magnitude, and geographic path. No missing values in most 
columns; `Path Width` and `Central Duration` are missing for ~4,200 partial eclipses, 
which don't have a central path by definition. Key columns for the dashboard: 
Calendar Date, Eclipse Type, Latitude/Longitude, and Eclipse Magnitude.
